DESAFIO 2: NUMERO PRIMO
Captura de tela 2026-04-23 134838.png

```python
import math # Esta linha importa o módulo 'math', que contém funções matemáticas. Ele será usado para calcular a raiz quadrada.

def eh_primo(n):
  """Verifica se um número é primo.

  Args:
    n: O número inteiro a ser verificado.

  Returns:
    True se o número for primo, False caso contrário.
  """
  if n <= 1: # Esta linha verifica se o número 'n' é menor ou igual a 1.
    return False  # Se 'n' for menor ou igual a 1, a função retorna 'False', pois esses números não são primos por definição.
  # As próximas linhas (comentários) explicam a otimização usada para verificar a primalidade.
  # Otimização: para verificar se um número é primo, basta testar divisores até a sua raiz quadrada.
  # Se não encontrar um divisor até lá, o número é primo.
  for i in range(2, math.isqrt(n) + 1): # Esta linha inicia um loop 'for'. Ele itera de 2 até a raiz quadrada inteira de 'n' (incluindo a raiz), verificando cada 'i' como um potencial divisor.
    if n % i == 0: # Dentro do loop, esta linha verifica se 'n' é divisível por 'i' (ou seja, se o resto da divisão é 0).
      return False  # Se 'n' for divisível por 'i', significa que 'n' tem um divisor além de 1 e ele mesmo, então a função retorna 'False' (não é primo).
  return True  # Se o loop terminar sem encontrar nenhum divisor (ou seja, 'n' não foi divisível por nenhum 'i' testado), a função retorna 'True', indicando que 'n' é um número primo.
```

```python
def inverte_texto(texto):
  """Inverte uma string.

  Args:
    texto: A string a ser invertida.

  Returns:
    A string invertida.
  """
  return texto[::-1]
```
